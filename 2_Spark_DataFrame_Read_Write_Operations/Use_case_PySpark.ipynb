```python
%sql
-- drop catalog if exists cities cascade;
-- Creating Catalog With Default 
create catalog if not exists cities;
create schema if not exists cities.default;
create table if not exists cities.default.madurais_table(id int, name varchar(30), age int);
insert into cities.default.madurais_table values (1, 'Imran', 20), (2, 'Syed', 12), (3, 'Raees', 15);
-- Creating Catalog with schema 
create catalog if not exists Sf_catalogs;
create schema if not exists Sf_catalogs.schema;
create table if not exists Sf_catalogs.schema.salesforce_table(id int, name varchar(30), age int, dob date);
insert into Sf_catalogs.schema.salesforce_table values (1, 'Syed', 12, to_date('02111993', 'ddMMyyyy')), (2, 'Imran', 20, to_date('02111993', 'ddMMyyyy')), (3, 'Raees', 15, to_date('02111993', 'ddMMyyyy'));
-- Drop or Delete the table, schema or default, catalog
drop table if exists sf_catalogs.schema.salesforce_table;
drop schema if exists sf_catalogs.schema;
drop catalog if exists sf_catalogs cascade;
```

```python
# Custom schema with non familiar and familiar types
# Custom schema with non Familiar 
df = spark.read.format("csv").options(header= True).load("/Volumes/workspace/default/de_data_lake/sample/acc.csv");
print(df.show(5))
custom_schema_non_familiar = "Name string, Type string, BillingStreet string, BillingCity string, BillingState string, ShippingStreet string, Phone string";
print(custom_schema_non_familiar)
df = spark.read.format("csv").options(header = True).schema(custom_schema_non_familiar).load("/Volumes/workspace/default/de_data_lake/sample/acc.csv")
print(df.show(10))
# Custom Schema with Familiar
from pyspark.sql.types import StructType, StructField, StringType, IntegerType;
custom_schema_familiar =  StructType([
    StructField("Name", StringType(), False),
    StructField("Type", StringType(), True),
    StructField("BillingStreet", StringType(), True),
    StructField("BillingCity", StringType(), True),
    StructField("BillingState", StringType(), True),
    StructField("ShippingStreet", StringType(), True),
    StructField("Phone", StringType(), True)
])
df = spark.read.format("csv").options(header = True).schema(custom_schema_familiar).load("/Volumes/workspace/default/de_data_lake/sample/acc.csv")
display(df.show(10))
```
