```python
%sql
-- drop catalog if exists cities cascade;
-- Creating Catalog With Default 
create catalog if not exists cities;
create schema if not exists cities.default;
create table if not exists cities.default.madurais_table(id int, name varchar(30), age int);
insert into cities.default.madurais_table values (1, 'Imran', 20), (2, 'Syed', 12), (3, 'Raees', 15);
-- Creating Catalog with schema 
create catalog if not exists Sf_catalogs;
create schema if not exists Sf_catalogs.schema;
create table if not exists Sf_catalogs.schema.salesforce_table(id int, name varchar(30), age int, dob date);
insert into Sf_catalogs.schema.salesforce_table values (1, 'Syed', 12, to_date('02111993', 'ddMMyyyy')), (2, 'Imran', 20, to_date('02111993', 'ddMMyyyy')), (3, 'Raees', 15, to_date('02111993', 'ddMMyyyy'));
-- Drop or Delete the table, schema or default, catalog
drop table if exists sf_catalogs.schema.salesforce_table;
drop schema if exists sf_catalogs.schema;
drop catalog if exists sf_catalogs cascade;
```
